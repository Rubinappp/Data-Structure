After object is initalized sucessfully
the variable user1 will have reference to that new object.

```python
class User:
    def __init__(self,username,name,email):
        self.username=username
        self.name=name
        self.email=email
        print("User created")
    def introduce_yourself(self,guest_name):
        print(f"Hi! {guest_name } I'm {self.name}.Contact me at {self.email}")
    def __repr__(self):
        return f"username={self.username} name={self.name} email={self.email}"
    def __str__(self):
        return self.__repr__()
```

Objects do NOT store the string representation.so user1 stores only a reference to the User object.

When print(user1) is executed:
1. Python looks for __str__().
2. If found, it calls __str__().
3. __str__() calls self.__repr__() and return strings
4. __str__() returns same string to print function.
54. print() now will prints that string.



#### whats the purpose of defining __str__ and __repr__ within a class.?

user1 points to a User object. When we do print(user1), Python cannot automatically know how we want that object to be displayed. Therefore, Python calls the object's __str__() method. The string returned by __str__() is what gets printed.

### 4.Implementation

```python
class User:
    def __init__(self,username,name,email):
        self.username=username
        self.name=name
        self.email=email
        print("User created")
    def introduce_yourself(self,guest_name):
        print(f"Hi! {guest_name } I'm {self.name}.Contact me at {self.email}")
    
```

```python
rubina=User('rubina','Rubina Poudel','rubina@gmail.com')
bina=User('bina','bina Poudel','bina@gmail.com')
ina=User('ina','ina Poudel','ina@gmail.com')
nishu=User('nishu','nishu basnet','nishu@gmail.com')
```

```python
class UserDatabase:
    def __init__(self):
        self.users=[]
    def insert(self,user):
        i=0
        while i<len(self.users):
            if(self.users[i].username>user.username):
                break
            i=i+1
        self.users.insert(i,user)
    def find(self,username):
       
        for user in self.users:
            if(user.username == username):
                return user
    def update(self,user):
        target=self.find(user.username)
        target.name=user.name
        target.email=user.email
    def list_all(self):
        return self.users
        
        

        
```

```python
database=UserDatabase()
```

```python
database.insert(rubina)
```

```python
database.insert(nishu)
database.insert(bina)
```
