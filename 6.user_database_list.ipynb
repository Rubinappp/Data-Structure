### 4.Implementation

```python
class User:
    def __init__(self,username,name,email):
        self.username=username
        self.name=name
        self.email=email 
    def __repr__(self):
        return f"username={self.username} name={self.name} email={self.email}"
    def __str__(self):
        return self.__repr__()
```

```python
rubina=User('rubina','Rubina Poudel','rubina@gmail.com')
bina=User('bina','bina Poudel','bina@gmail.com')
ina=User('ina','ina Poudel','ina@gmail.com')
nishu=User('nishu','nishu basnet','nishu@gmail.com')
```

```python
class UserDatabase:
    def __init__(self):
        self.users=[]
    def insert(self,user):
        i=0
        while i<len(self.users):
            if(self.users[i].username>user.username):
                break
            i=i+1
        self.users.insert(i,user)
    def find(self,username):
       
        for user in self.users:
            if(user.username == username):
                return user
    def update(self,user):
        target=self.find(user.username)
        target.name=user.name
        target.email=user.email
    def list_all(self):
        return self.users
        
        

        
```

```python
database=UserDatabase()
```

```python
database.insert(rubina)
database.insert(nishu)
database.insert(bina)
```

```python
user=database.find('bina')
user
```

Suppose my backend is running on a machine with 8 CPU cores.

If each request requires about 10 seconds of CPU processing, then all 8 cores can become busy processing 8 requests simultaneously.

If a 9th arrives while all 8 cores are busy, it cannot start executing immediately. It must wait until one of the cores becomes available.

To solve it we can add more server,more core,more RAM but upgrading infrastructure  will cost millions of dollar for company.So we must choose efficient  data structure and algorithm.


#### Data Structure:
A way of organizing and storing data efficiently.
eg: Book shelf (list,dictionary,binary tree,hash table)

#### Algorithm:
A step-by-step procedure used to solve a problem or perform an operation on that data.eg: way to find  book in bookshelf.(linear search,binary search)

#### Conclusion
An efficient data structure combined with an efficient algorithm leads to an efficient program.

Examples:

1. List (unsorted) + Linear Search → O(n)

2. Sorted List + Binary Search → O(log n)

3. Binary Search Tree + BST Search → O(log n) average

The algorithm must match the properties of the data structure. Otherwise, we lose efficiency or the algorithm may not work correctly.

When solving a problem, I should ask:

1. Which data structure is most suitable for storing my data?

2. Which algorithm is designed to work efficiently with that data structure?

3. Does the algorithm take advantage of the properties of the data structure?
